```python
# Cell 1: Import Library dan Load Dataset
import pandas as pd

# Memuat dataset
file_path = 'top_insta_influencers_data[1].csv'
df = pd.read_csv(file_path)

print("Dataset berhasil dimuat!")
```

```python
# Cell 2: Memeriksa Jumlah Baris dan Kolom
rows, cols = df.shape
print(f"Jumlah Baris : {rows}")
print(f"Jumlah Kolom : {cols}")
```

```python
# Cell 3: Memeriksa Kolom dan Tipe Data
print("=== Ringkasan Kolom dan Tipe Data ===")
df.info()
```

```python
# Cell 4: Memeriksa Data Kosong (Missing Values)
missing_data = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    'Jumlah Data Kosong': missing_data,
    'Persentase (%)': missing_percentage.round(2)
})

print("=== Analisis Data Kosong ===")
print(missing_df)
```

```python
# Cell 5: Menampilkan Sampel Data
print("=== 5 Baris Pertama Data ===")
display(df.head())

print("=== 5 Baris Terakhir Data ===")
display(df.tail())
```
